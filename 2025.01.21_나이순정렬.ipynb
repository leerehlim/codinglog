### 문제 요구 사항 
이 문제는 사용자로부터 나이와 이름 정보를 입력받아 리스트에 저장하고, '나이로만' 기준으로 정렬하여 출력하는 기능이다.
### 풀이과정 
sort() ,sorted() 함수를 사용하여 효과적으로 정렬을 수행 
list 안에 첫번째로 위치하는 나이를 기준으로 정렬을 수행
### 잘한 부분
- 문제를 이해하고 이에 대한 풀이과정을 simulation한 것 
- sort 뿐만 아니라, sorted 에 대해서도 코드를 구현 
### 알게 된 부분 
- sort , sorted의 기능 
- sort() 함수는 기본적으로 입력한대로 순서가 정렬된다. 
### 향후 계획 
    silver 정렬 문제 마무리 

```python
# import sys 
# input = sys.stdin.readline
N = int(input())
mem_lst = list()
for _ in range(N): 
    age, name = input().split()
    mem_lst.append([int(age),name])

mem_lst.sort(key = lambda x : x[0])

for i in mem_lst : 
    for real in i : 
        print(real, end = ' ')
    print()
```

```python
# 약간 변형 
N = int(input())
member = list()

for _ in range(N):
    age, name = input().split()
    member.append([int(age), name])

for i in sorted(member, key = lambda x : x[0]):
    print(i[0], i[1])
    
   
    
```
