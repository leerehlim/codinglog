### 문제 요구 사항 
들어온 문자열의 글자수를 세서 작은 순서대로 다시 재정렬하는 것
### 풀이과정 

### 잘한 부분
- 문제를 이해하고 이에 대한 풀이과정을 simulation한 것 
- clean code 라고 여길 수 있는 다른 코드를 찾아본 것 
### 알게 된 부분 
- sort 의 원리
- sort() 함수를 쓰면 사전적으로 정렬된다. 
- 글자 수를 기준으로 정렬할때는 len() 을 key 로 잡아준다. 
### 향후 계획 
    silver 정렬 문제를 푼다. 

```python
import sys
input = sys.stdin.readline
N = int(input().strip()) # 혹시 모르니 양옆의 공백 짜르기 
str_lst = list() 

for _ in range(N): 
    st_r = str(input().strip()) # 혹시 모르니 양옆의 공백 짜르기 
    if st_r not in str_lst :   
        str_lst.append(st_r)

str_lst.sort()
str_lst.sort(key = len)

for word in str_lst:
    print(word) 
```
