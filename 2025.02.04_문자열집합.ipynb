### 문제 요구 사항 
- 문자열 집합 vs 탐색해야 할 문자열 집합을 대조해서 몇개인지 count 하는 것이 아닐지 

### 풀이과정 
S에 들어갈 문자열들을 List로 만들어 집합화해준다. 
후에 M개의 문자를 반복문에 넣고 바로바로 list에 있는지 체크하면서 있다면 count + 1 하는 방식으로 코드를 짰다. 

### 잘한 부분 및 개선할 부분 
- 문제를 이해하고 어떤 방식으로 풀지에 대해서 설계를 할 수 있었음. 
- 더 효율적인 방식으로 코드를 짜는 습관을 기를 필요가 있음.

### 향후 계획 
- silver 문제 독파!!    

```python
import sys
input = sys.stdin.realine

n,m = map(int, input().split())

str_set = list() 
cnt = 0

for _ in range(n):
    word = str(input())
    str_set.append(word)

for _ in range(m):
    word2 = str(input())
    if word2 in str_set:
        cnt += 1 
    

    
print(cnt)


### 런타임 에러 
### 두번의 반복문 없이 바로바로 체크해주는 방식으로 
```
