### 알게 된 부분 
- 풀이 방식은 의외로 단순하다. 더 많은 문제를 풀어봄으로써 컴퓨터적인 사고를 해야 한다는 것 
- dictinary 와 set을 이용해서 데이터를 변형시키는 방법을 배울 수 있었다. 
- set 은 데이터의 요소 중복값을 없애준다는것을 기억하자 그러나 순서는 장담할 수 없다. 
- dictionary[key] = value 라는 것 ! 잘 이용해야 한다. 
### 향후 계획 
    정렬문제 한번 skimming 및 정리 필요 

```python
import sys 
input = sys.stdin.readline
N = int(input())
num_lst = list(map(int, input().split())) 
 

num_sort = sorted(list(set(num_lst)))
dic = {}
for i in range(len(num_sort)): 
    dic[num_sort[i]] = i # key - value  

for i in num_lst : 
    print(dic[i], end = ' ')
    
```
