### 문제 요구 사항 
좌표 정렬하기 2 는 x가 아닌 y를 기준으로 2차원 배열을 정렬하는 것이 문제의 핵심이다. 
### 풀이과정 
- 어제 풀었던 2차원 배열 sort 에 대해 숙지하여 오늘 스스로 문제를 풀어냈다. 
### 잘한 부분
- sort 를 2차원 배열 내에서 사용할 때 어떤 방식으로 sorting 할 수 있는지에 대해 알게 되었다. 
### 알게 된 부분 
- 정렬 부분에 있어서 다양한 자료구조에 있어 어떤 방식으로 정렬할 수 있는지 더 많은 문제를 섭렵 해보면 좋을 것 같음. 
### 향후 계획 
    내일도 백준의 정렬 문제를 풀어보자. silver 이상 


```python
import sys 

N = int(sys.stdin.readline())
arr = list()
for _ in range(N):
    x_i, y_i = map(int, sys.stdin.readline().split())
    arr.append([x_i, y_i]) # append 하여 2차원 배열 구조를 만들어준다. 


arr.sort(key = lambda x : (x[1], x[0]))

for i in arr :
    print(i[0], i[1])
```


## 2차원 배열 정렬에서 x, y 기준을 정하는 방법
lst = [[2, 1], [3, 4], [1, 2], [1, 3], [3, 2]] 
lst.sort(key = lambda x : x[0]) 으로 하는 경우 
- [[1, 2], [1, 3], [2, 1], [3, 4], [3, 2]] 이런방식으로 x 값에 대해 먼저 정렬을 한다. 

### case 2 
만약 x에 대해서는 오름차순으로 정렬을 하고 , y는 내림차순으로 정렬하는 경우에는 key 를 지정해서 기준을 세워주면 되는데 
lst.sort(key = lambda x: x[0], -x[1]) # -를 붙여주면 역방향으로 정렬함을 의미하는 것

### case 3 
만일 y를 기준으로 오름차순 한다고 가정하면 이번에도, 
lst.sort(key = lambda x : x[1], x[0]) 이런식으로 가정하여 정렬을 할 수 있음.
