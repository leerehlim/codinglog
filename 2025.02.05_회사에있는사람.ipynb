### 문제 요구 사항 
enter, leave 를 식별함으로써 리스트에 남은 사람을 사전 순으로 정렬(역사전) 
### 풀이과정 
enter , leave 를 보고 enter일 경우 append로 list 에 넣는다. 
leave 가 뜰 경우 이미 들어가 있는 정보일테니 요소 삭제 - pop이나 remove를 이용? 
남은 리스트 sort (역 정렬)한다. 
### 잘한 부분 및 개선할 부분 
문제 이해력이 이전보다 좋아지고 있음을 느낀다. 
해시맵으로 이용하기 전에 코드로 test 는 맞았으나, 시간 복잡도 측면에서 놓쳤다. 해시맵 구조의 경우 dictionary 를 이용하는 것이 입출력 부분에서 상당한 시간 리소스를 확보해주기 때문에 향후 이부분을 인지하여 코드 작성하는 것을 익힌다! 
### 알게 된 부분 
해시맵 코드 구현 방안
string - strip 기능사용방법
"\n".join(리스트) 사용방법
### 향후 계획

```python
# 1차 시도 -> 시간 초과 
import sys 
input = sys.stdin.readline()

n = int(input())
n_lst = list() 
for _ in range(n):
    name, status = map(str, input().split())
    if status == 'enter': 
        n_lst.append(name)
    if status == 'leave':
        n_lst.remove(name)
n_lst.sort(reverse=True)

for name in n_lst:
    print(name)
print()


        
```

```python
# 2차 시도 -> 시간초과 
import sys 
input = sys.stdin.readline

n = int(input())
n_lst = list() 
for _ in range(n):
    name, status = input().split()
    if status == 'enter': 
        n_lst.append(name)
    if status == 'leave':
        n_lst.remove(name)
n_lst.sort(reverse=True)

for name in n_lst:
    print(name)
print()
```

```python
# 시간초과로 해결할 수 없다. 
# 해시맵 구조를 이용한다. 해시맵 구조란 ? Dictionary 를 이용한다는것! 
# 해시맵을 통해 key - value 값으로 저장되며 추가 삭제연산이 매우 빠르다. 
```

```python
# 3차 시도 -> 맞았습니다!!
import sys 
input = sys.stdin.readline 

n = int(input())
company = {}  # key - value , 이름 - 출입 상황 
for _ in range(n):
    name , status = input().strip().split() 
    if status == 'enter': 
        company[name] = True 
    else :
        del company[name]


print("\n".join(sorted(company.keys(), reverse =True)))

      
```

# Python 의 String 의 내장함수를 활용한 문자제거 

- strip([chars]) : 인자로 전달된 문자를 String 의 왼쪽과 오른쪽에서 제거한다.
- lstrip([chars]) : 인자로 전달된 문자를 String 의 왼쪽에서 제거한다. 
- rstrip([chars]) : 인자로 전달된 문자를 String 의 오른쪽에서 제거한다.
